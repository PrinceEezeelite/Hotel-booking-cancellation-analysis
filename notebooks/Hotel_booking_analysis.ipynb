## DATA VALIDATION

### Removing where adults,children,babies = 0

```python
df = df[-((df['adults'] == 0) &
         (df['children'] == 0) &
         (df['babies'] == 0))]
```

### Outlier Detection (ADR = PRICE)

```python
df['adr'].describe()
```

```python
sns.boxplot(x=df['adr'])
```

```python
df = df[df['adr'] < 500]
df =df[df['adr'] >= 0]
```

```python
sns.boxplot(x=df['adr'])
```

## FEATURE ENGINEERING 

### Total Nights

```python
df['total_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
```

### Total Guests

```python
df['total_guests'] = df['adults'] + df['children'] + df['babies']
```

### Family's 

```python
df['is_family']= (df['total_guests'] > 1).astype(int)
```

### Booking Changes Indicator

```python
df['has_change'] = (df['booking_changes'] > 0).astype(int)
```
