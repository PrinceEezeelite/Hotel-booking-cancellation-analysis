## DATA VALIDATION

### Removing where adults,children,babies = 0

```python
df = df[-((df['adults'] == 0) &
         (df['children'] == 0) &
         (df['babies'] == 0))]
```

### Outlier Detection (ADR = PRICE)

```python
df['adr'].describe()
```

```python
sns.boxplot(x=df['adr'])
```

```python
df = df[df['adr'] < 500]
df =df[df['adr'] >= 0]
```

```python
sns.boxplot(x=df['adr'])
```

## FEATURE ENGINEERING 

### Total Nights

```python
df['total_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
```

### Total Guests

```python
df['total_guests'] = df['adults'] + df['children'] + df['babies']
```

### Family's 

```python
df['is_family']= (df['total_guests'] > 1).astype(int)
```

### Booking Changes Indicator

```python
df['has_change'] = (df['booking_changes'] > 0).astype(int)
```

## WEEK 2: EXPLORATORY DATA ANALYSIS (EDA)

## Cancellation Rate

```python
df['is_canceled'].value_counts(normalize=True)
```

### Visual:

```python
sns.countplot(x='is_canceled', data=df)
plt.title("Cancellation Distribution")
plt.show()
```

#### Cancellation Distribution
- Majority of bookings are not canceled (~70%+)
- A substantial portion (~25–30%) still gets canceled
- Indicates cancellation is a meaningful operational issue

#### Business Insight:
While most customers follow through, the cancellation rate is high enough to impact revenue. Implement stricter policies (e.g., partial prepayment, flexible rescheduling) to reduce losses.

## LEAD TIME vs CANCELLATION

```python
sns.boxplot(x='is_canceled', y='lead_time', data=df)
plt.show()
```

#### Lead Time vs Cancellation
- Customers who cancel tend to have longer lead times
- Median lead time is noticeably higher for canceled bookings
- More variability and extreme values (outliers) in canceled group

#### Business Insight:
The earlier a booking is made, the higher the risk of cancellation. Introducing incentives for shorter lead-time bookings or stricter policies for early bookings can help reduce cancellations.

## ADR (PRICE) ANALYSIS

```python
sns.boxplot(x='is_canceled', y='adr', data=df)
plt.show()
```

#### ADR (Price) Analysis
- Canceled bookings have slightly higher median ADR
- Wider spread in ADR among canceled bookings
- High-price bookings show more volatility

#### Business Insight:
Higher-priced bookings are more prone to cancellation. Consider offering flexible pricing tiers or value-added services to justify higher rates and reduce cancellation likelihood.

## SEASONAL DEMAND

```python
sns.countplot(x='arrival_date_month', data=df,
              order=df['arrival_date_month'].value_counts().index)
plt.xticks(rotation=45)
plt.show()
```

#### Seasonal Demand
- Peak demand occurs in August and July
- Lowest demand seen in January, November, and December
- Clear seasonal booking pattern

#### Business Insight:
Leverage peak seasons with premium pricing strategies, and boost off-season demand through promotions, discounts, or bundled offers.

## LEAD TIME BUCKET ANALYSIS

### Create buckets:

```python
df['lead_time_group'] = pd.cut(df['lead_time'],
                              bins=[0, 30, 90, 180, 365, 1000],
                              labels=['0-30', '31-90', '91-180', '181-365', '365+'])
```

### Plot:

```python
sns.barplot(x='lead_time_group', y='is_canceled', data=df)
plt.title("Cancellation Rate by Lead Time Group")
plt.show()
```

#### Lead Time vs Cancellation
- Cancellation rate increases as lead time increases
- Short lead time (0–30 days) has the lowest cancellation (~18%)
- Long lead time (365+ days) has the highest cancellation (~40%+)
- Clear upward trend between booking time and cancellation risk

#### Business Insight:
The earlier customers book, the more likely they are to cancel. Implementing stricter policies (e.g., partial prepayment or penalties) for long lead-time bookings can reduce cancellations.

## ADR (PRICE) vs DEMAND

```python
sns.scatterplot(x='adr', y='total_nights', hue='is_canceled', data=df)
plt.show()
```

#### ADR (Price) vs Demand
- Most bookings cluster at lower ADR values (₦0–₦150 range) with short stays
- As ADR increases, total nights tend to decrease (inverse relationship)
- Cancellations (orange points) are spread across all price ranges but slightly more frequent at mid-to-high ADR levels

#### Business Insight:
Higher pricing does not necessarily drive longer stays and may slightly increase cancellation risk. Consider optimizing pricing strategies (e.g., discounts for longer stays) to boost both retention and occupancy.

## REPEATED VS NEW CUSTOMERS

#### Repeated vs New Customers

## LENGTH OF STAY VS CANCELLATION

```python
sns.boxplot(x='is_canceled', y='total_nights', data=df)
plt.show()
```

####  Length of Stay vs Cancellation

- Most bookings are short stays (1–5 nights)
- The median stay length is nearly identical for both canceled and non-canceled bookings
- Both groups show similar distributions with overlapping ranges
- Presence of outliers indicates some unusually long stays, but they do not significantly impact cancellation patterns

#### Business Insight:
Length of stay is not a strong predictor of cancellations. Hotels should focus more on factors like lead time, deposit type, and customer engagement when developing strategies to reduce cancellation rates.
